```python
# Import packages
import pandas as pd
import requests # https://requests.readthedocs.io/en/latest/
from bs4 import BeautifulSoup # https://beautiful-soup-4.readthedocs.io/en/latest/
```

```python
# API pull data from Mountain project

# This URL is currently set by going into Mountain Project, finding the specific
# area, searching for routes, and setting the filters to find all boulder probs.
# I can export a csv of the table on this page, but scraping it would be more
# ideal since it would automate this step.
# If I go the csv route, info on cleaning the data is here:
# https://github.com/jnd18/mountain-project-eda/blob/master/rrg-and-nrg.ipynb

URL = "https://www.mountainproject.com/route-finder?diffMaxaid=75260&diffMaxboulder=21700&diffMaxice=38500&diffMaxmixed=65050&diffMaxrock=5500&diffMinaid=70000&diffMinboulder=20000&diffMinice=30000&diffMinmixed=50000&diffMinrock=1800&is_sport_climb=1&is_top_rope=1&is_trad_climb=1&pitches=0&selectedIds=105716862&sort1=area&sort2=rating&stars=0&type=boulder&viewAll=1"

page = requests.get(URL) # Step 1: get the page content
soup = BeautifulSoup(page.text, 'html.parser') # Step 2: parse the content

# Step 3: extract data
print(soup.title.text)
```
